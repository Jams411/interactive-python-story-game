```python
import time

def say(text):
    print(text)
    time.sleep(2)

# 🎤 Intro
print("🎬 Welcome to the WORST 5 minutes of someone’s life.")
print("⚠️ Rule: The worse your answers, the better this gets.\n")

# 😂 Funny prompts before each input
print("Alright… who is today's victim? 😈")
name = input("👤 Enter a name (friend, celeb, professor… be bold): ")

print("\nWe need a smell that should be illegal...")
gross_thing = input("🤢 Something disgusting (e.g., wet socks, expired milk, gym locker): ")

print("\nWhere will this disaster happen?")
place = input("📍 Enter a place (classroom, subway, wedding, job interview): ")

print("\nTime for public humiliation...")
action = input("😬 Embarrassing action (tripped, farted loudly, sneezed on someone): ")

print("\nFinal touch… something they should NEVER eat.")
food = input("🍜 Weird food (pickle ice cream, banana pizza, chocolate + ketchup): ")

print("\n⏳ Generating disaster...\n")
time.sleep(2)

# 🎭 Story begins
print("\n--- THE WORST 5 MINUTES OF YOUR LIFE ---\n")

say(f"{name} walked into the {place}, feeling confident.")
say("Too confident.")

say("Suddenly… a smell hit the room.")
say(f"Someone whispered: 'What is that? Smells like {gross_thing}.'")

say(f"{name} froze.")
say(f"{name}: 'That’s… not me.'")

say("Everyone stared.")

say(f"A random kid stood up and yelled: 'IT'S {name}!'")
say("Chaos.")

say(f"In panic, {name} tried to act normal… and accidentally {action}.")
say("Dead silence.")

say("Then someone in the back made a noise:")
say('"BRUUUUH."')

say(f"Trying to recover, {name} grabbed some {food} and took a big bite.")
say("Instant regret.")

say(f"{name}: 'Why did I do that…'")
say("Teacher walks in. Looks around. Leaves immediately.")

say(f"At that moment, {name} accepted their fate.")

say("Reputation: destroyed.")
say("Dignity: gone.")
say("Legend status: unlocked.")

say("\nTHE END.\n")
```
